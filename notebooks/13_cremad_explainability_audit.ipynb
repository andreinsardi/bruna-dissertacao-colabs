# Explainability evaluation for CREMA-D

This notebook measures Integrated Gradients completeness and block attribution, actor-level ADI under race-label permutation, and temporal deletion against a random order. It uses the fixed-test EmoT checkpoint and the output of the parameter-randomization evaluation. The checkpoint and data should come from the same CREMA-D run.

```
from pathlib import Path
import subprocess
import sys

from google.colab import drive

drive.mount('/content/drive')
base_dir = Path('/content/drive/MyDrive/cremad_fairness_xai')
output_dir = base_dir / 'explainability_audit'
repo = Path('/content/bruna-dissertacao-colabs-v1.2.0')
if not repo.exists():
    subprocess.run([
        'git', 'clone', '--depth', '1', '--branch', 'v1.2.0',
        'https://github.com/andreinsardi/bruna-dissertacao-colabs.git', str(repo)
    ], check=True)
```

```
# Use the same checkpoint and parameter-randomization outputs as the dissertation.
subprocess.run([
    sys.executable, str(repo / 'scripts' / 'evaluate_explainability.py'),
    '--base-dir', str(base_dir), '--output-dir', str(output_dir)
], check=True)
```
